```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x,A,w):
    new_list = []
    for j in range(len(x)):
        y = A * np.sin(w*x[j])
        new_list.append(y)
    new_array = np.array(new_list)
    return new_array

z = np.random.randint(0,20, (1,3))

print(z)

make_sine_wave(z, 2, 1)
```

```python
def make_sine_wave(x,A,w):
    new_list = []
    for j in range(len(x)):
        y = A * np.sin(w*x[j])
        new_list.append(y)
    new_array = np.array(new_list)
    return new_array
    
fig, ax = plt.subplots(1,3, figsize=(25,5))

amplitude = [0.5, 1, 1.5, 2, 2.5]

frequency = [1,2,3,4,5]

a = zip(amplitude, frequency)

for i in a:
    x = np.linspace(0,2*(np.pi), 1000)
    y = make_sine_wave(x, i[0], 1)
    ax[0].plot(x,y)
    ax[0].set_title('Variation in Amplitude', fontsize=20)
    ax[0].set_xlabel('Time', fontsize=20)
    ax[0].set_ylabel('Displacement', fontsize=20)

b = zip(amplitude, frequency)

for i in b:
    x = np.linspace(0,2*(np.pi), 1000)
    y = make_sine_wave(x, 1, i[1])
    ax[1].plot(x,y)
    ax[1].set_title('Variation in Frequency', fontsize=20)
    ax[1].set_xlabel('Time', fontsize=20)
    ax[1].set_ylabel('Displacement', fontsize=20)

c = zip(amplitude, frequency)

for i in c:
    x = np.linspace(0,2*(np.pi), 1000)
    y = make_sine_wave(x, i[0], i[1])
    ax[2].plot(x,y)
    ax[2].set_title('Variation in Amplitude and Frequency', fontsize=20)
    ax[2].set_xlabel('Time', fontsize=20)
    ax[2].set_ylabel('Displacement', fontsize=20)

plt.show()
```
