```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
array = np.loadtxt("C:/Users/Anirvan Gautam/python_decal_fa25/course_assignments/homework7/star_data.txt", skiprows=1)
array
```

```python
np.shape(array)

#There are 2642 rows!!!
```

```python
magnitude = array[:, 0]

print(magnitude.max()) #The 264th star is the dimmest star

# print(np.where(array == 12.49))

print(magnitude.min()) #The 1988th star is the brightest

# print(np.where(array == 2.24))

temperature = array[:, 1]

print(temperature.max()) # The 2613th star is the hottest

# print(np.where(array == 10112.78057))

print(temperature.min()) # The 1923rd star is the coolest

# print(np.where(array == 3653.20966))
```

```python
array[264][1]
```

```python
brightest = array[264]

dimmest = array[1988]

hottest = array[2613]

coolest = array[1923]

log_magnitude = np.log(magnitude)

log_temperature = np.log(temperature)

plt.figure(figsize=(8,6))

plt.scatter(log_temperature, log_magnitude)

plt.scatter(np.log(brightest[1]), np.log(brightest[0]), color="black", marker="*", s=300, label="Brightest")

plt.scatter(np.log(dimmest[1]), np.log(dimmest[0]), color="purple", marker="*", s=300, label="Dimmest")

plt.scatter(np.log(hottest[1]), np.log(hottest[0]), color="green", marker="*", s=300, label="Hottest")

plt.scatter(np.log(coolest[1]), np.log(coolest[0]), color="orange", marker="*", s=300, label="Coolest")

plt.title("Hertzsprung Russell Diagram")

plt.xlabel("Temperature (log K)")

plt.ylabel("Magnitude")

plt.gca().invert_xaxis()

plt.gca().invert_yaxis()

plt.legend(loc="lower left")

plt.show()
```
