```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy import units as u
from astropy import constants as const
```

```python
df = pd.read_csv("~/python_decal_fa25/course_assignments/homework9/solar_system.csv")
df
```

```python
df1 = df.set_index("Attribute").T
df1.index.name = "Planet"
df1.reset_index(inplace=True)
df1.columns.name = None

df1
```

```python
print(df.shape)

print(df1.shape)
```

```python
"""
The shape before transposing = (20,11)

The shape after = (10,21)

The 1st line of code defines a new matrix where the previous matrix has been transposed.

It also makes the indexes for the new columns the attributes of the planets

The 2nd line of code sets the name of the index column as "Planets", since they are in fact the planets

The 3rd line of code retrieves the numbered index for the rows.

The 4th line of code removes the descriptors for the columns.

The new dataframe isn't 11 by 20 because we took the row naming all the attributes and made it the index in the new dataframe. So we will naturally

have 1 less row. We also made the planets their own column instead of an index, so we will naturally have 1 more column.
"""
```

```python
for i in df1.columns:
    print(i)
```

```python
units = 0
columns_with_units = []
columns_without_units = []
no_units = 0
for i in df1.columns:
    if ("(" in i) == True:
        units += 1
        columns_with_units.append(i)
    else:
        no_units += 1
        columns_without_units.append(i)

print(units)
print(columns_with_units)
print(no_units)
print(columns_without_units)
```

```python
"""
The actual values are all strings. This is problematic for data analysis because we can't perform basic math operations (i.e mean) on strings.

"""
```

```python
df1.columns

for i in df1.columns:
    if (i != "Planet") & (i != "Ring System?") & (i != "Global Magnetic Field?"):
        df1[i] = pd.to_numeric(df1[i], errors="coerce")
```

```python
df1.dtypes
```

```python
for i in df1.columns:
    print(df1[i].apply(type))
```

```python
"""
The three different types of data are now floats, integers, and objects.

float64 is a 64-bit float

int64 is a 64-bit integer

object is anything from mixed data types to strings to complex objects like dictionaries.

Planet, Ring System and Magnetic Field are still objects because they were kept as strings.
"""
```

```python
unit = 1e6 * u.km
old_col = 'Distance from Sun (10^6 km)'
new_col = 'Distance from Sun (km)'

def attach_units(column):
    new_values = []
    for i in df1[column]:
        new_values.append(i * unit)
    # print(new_values)
    df1[column] = new_values
    # print(df1[column])
    df1.rename(columns={old_col: new_col}, inplace=True)

attach_units('Distance from Sun (10^6 km)')
```

```python
"""
An astronomical unit is the average distance from the Sun to the Earth
"""

const.au
```

```python
x= const.au

x.to(u.km)
```

```python
def converting_distances(column):
    new_val = []
    for i in df1[column]:
        j = i * u.km
        # print(j.to(u.yr))
        new_val.append(j.to(u.au))
    df1[column] = new_val

converting_distances('Diameter (km)')

df1.rename(columns={'Diameter (km)': 'Diameter (AU)'}, inplace=True)
```

```python
converting_distances('Distance from Sun (km)')

df1.rename(columns={'Distance from Sun (km)': 'Distance from Sun (AU)'}, inplace=True)
```

```python
converting_distances('Perihelion (10^6 km)')

df1.rename(columns={'Perihelion (10^6 km)': 'Perihelion (AU)'}, inplace=True)
```
