```python
import numpy as np
```

```python
x = np.median(wow['latitude'])# the median latitude is +39 degrees
x
```

```python
y = np.median(wow['longitude']) # the median longitude is -89 degrees
y
```

```python
wow['Latitude Deviation'] = abs(wow['latitude']-x)
wow
```

```python
wow['Longitude Deviation'] = abs(wow['longitude']-y)
wow
```

```python
print(wow.shape) # There are now 52 rows and 6 columns
```

```python
x = wow[wow['Latitude Deviation'] == min(wow['Latitude Deviation'])]

y = x['name'].tolist()

print(f"The states closest to the median latitude are {y} by {min(wow['Latitude Deviation']):.2f}")
```

```python
y= wow[wow['Latitude Deviation'] == max(wow['Latitude Deviation'])]

z = y['name'].tolist()

print(f"The state farthest from the median latitude is {z} by {max(wow['Latitude Deviation']):.2f}")
```

```python
z=wow[wow['Longitude Deviation'] == min(wow['Longitude Deviation'])]

a = z['name'].tolist()

print(f"The states closest to the median longitude are {a} by {min(wow['Longitude Deviation']):.2f}")
```

```python
a=wow[wow['Longitude Deviation'] == max(wow['Longitude Deviation'])]

b = a['name'].tolist()

print(f"The state farthest from the median latitude is {b} by {max(wow['Longitude Deviation']):.2f}")
```

```python
pd.pivot(wow, columns='Latitude Deviation', index='latitude', values='state') 

# I found this function under the data manipulations tab on the website.

# The function allows you to choose a column from your dataframe whose values will become columns in your new table. You can also choose how to index the

# entries in your new table based on a separate column for your dataframe, and what values show up in each cell. For example, I ran pd.pivot on wow 

# and chose the values from Latitude Deviation to be the columns in my new table. I indexed every entry from wow based on latitude and made it so that 

# the cell would show the state that satisfied the latitude and latitude deviation. pd.pivot() can help identify correlations in the data, as I've done

# here figuring out what latitude values for each state are giving me a specific latitude deviation.
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
datfram = sns.load_dataset('dowjones')
datfram
```

```python
# There are no columns with NaN values
```

```python
datfram.shape # the shape of my dataframe is 649 rows and 2 columns
```

```python
datfram.head()
```

```python
datfram.tail()
```

```python
# The first column name is the date that the price of the dow jones was measured.

# The second column is the price of the dow jones on that date.
```

```python
fig, ax = plt.subplots(2,2, figsize=(10,10))

plt.subplots_adjust(wspace=1, hspace=0.5)

x = datfram['Date']

y= datfram['Price']

ax[0,0].scatter(x,y, s=1, c='g')
ax[0,0].set_title("Dow Jones Industrial Average, 1914-1968", fontsize=10)
ax[0,0].set_xlabel("Year")
ax[0,0].set_ylabel("Price (Dollars)")
ax[0,0].legend(datfram)

ax[0,1].scatter(y,x, s=1, c='r')
ax[0,1].set_title("Dow Jones Industrial Average, 1914-1968 (flipped)",fontsize=10)
ax[0,1].set_ylabel("Year")
ax[0,1].set_xlabel("Price Dollars")

sns.histplot(y, ax=ax[1,0])
ax[1,0].set_title("Price Point Histogram")
ax[1,0].set_ylabel("Number of Occurences")
ax[1,0].set_xlabel("Price Point (Dollars)")

sns.boxplot(x)
ax[1,1].set_title("Distribution of years in the data")
ax[1,1].set_ylabel("Year")


plt.show()
```

```python
# The first subplot represents the growth of the Dow Jones' price from 1914 to 1968 month by month. The only columns to choose from were price and date. 
# The graph says a lot about the exponential growth of the American economy in the Post-War Era.

# The second subplot is the first subplot flipped. The columns are the same and the graph shows the frequency at which the Dow Jones was a certain
# price point.

# The thrid subplot is a histogram more clearly showing how many times the Dow Jones was within a certain price range. The subplot demonstrates that 
# the growth of the Dow Jones was rapid: the index was more often at lower price ranges in the 20th century than at higher price ranges.

# The fourth subplot is a very basic subplot which shows the distribution of years sampled in the data. Since the data points were sampled equally
# throughout the year from 1914-1968, all the dates fall nicely into the displayed boxplot.
```
