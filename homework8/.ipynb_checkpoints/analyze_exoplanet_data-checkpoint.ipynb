```python
# Planet Name
    # pl_name
    # Planet name most commonly used in the literature
    # Name of the Exoplanet

# print(ex['pl_name'])

# Host Name
    # hostname
    # Stellar name most commonly used in the literature
    # Name of the host star
    
# print(ex['hostname'])


# Discovery Method
    # discoverymethod
    # Method by which the planet was first identified
    # How the planet was first discovered

# print(ex['discoverymethod'])

# Spectral Type
    # st_spectrype
    # Classification of the star based on their spectral characteristics
    # following the Morgan-Keenan system
    # Type of star that hosts the exoplanet based on its temperature, spectral features, etc

# print(ex['st_spectype'])

# Equilibrium Temperature
    #  pl_eqt
    # The equilibrium temperature of the planet as modeled by a black body
    # heated only by its host star
    # The very ideal temperature measurement of the planet

# print(ex['pl_eqt'])

# Planet Mass*sin(i) [Jupiter Mass]
    # pl_bmassj
    # Best planet mass estimate available, in order of preference:
    # Mass, M*sin(i)/sin(i), or M*sin(i)
    # Mass of the planet adjusted for the inclination of the planet's orbit

# print(ex['pl_bmassj'])

ex_new = ex[['pl_name', 'hostname', 'discoverymethod', 'st_spectype', 'pl_eqt', 'pl_bmassj']]

ex_new
```

```python
need_to_remove = ex_new.loc[np.isnan(ex_new['pl_eqt'])==True]
```

```python
ex_new_new = ex_new.drop(need_to_remove.index)
```

```python
need_to_remove2 = ex_new_new.loc[np.isnan(ex_new['pl_bmassj'])==True]

ex_new_new_new = ex_new_new.drop(need_to_remove2.index)

ex_new_new_new
```

```python
need_to_remove3 = ex_new_new_new[ex_new_new_new['st_spectype'].isna()]

ex_final = ex_new_new_new.drop(need_to_remove3.index)

ex_final
```

```python
# Transit, Radial Velocity and Imaging

# There are 52 transits, 13 Radial Velocities, and 2 Imaging exoplanets

# Transit method is the discovery of an exoplanet based on evidence of transit:
# a dip in the light curve of its host star

# Radial Velocity method is the detection of an exoplanet not by seeing it, but by noticing
# the spectroscopic wobble in its host star

# Imaging is just getting a direct image of the exoplanet
```

```python
plt.figure(figsize=(8,6))

plt.title("2019 Exoplanet Discoveries")

plt.xlabel("Mass Estimate (Jupiter Masses)")

plt.ylabel("Ideal Equilibrium Temperature")

disc_method_list = ex_final['discoverymethod'].unique()
# print(disc_method_list)

for i in disc_method_list:
    disc_method_subset = ex_final[ex_final['discoverymethod'] == i]
    plt.scatter(disc_method_subset['pl_bmassj'], disc_method_subset['pl_eqt'], label=f"{i}")

plt.legend()

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.title("2019 Exoplanet Discoveries")

plt.xlabel("Mass Estimate (Jupiter Masses)")

plt.ylabel("Ideal Equilibrium Temperature")

disc_method_list = ex_final['discoverymethod'].unique()
# print(disc_method_list)

for i in disc_method_list:
    disc_method_subset = ex_final[ex_final['discoverymethod'] == i]
    plt.scatter(np.log(disc_method_subset['pl_bmassj']), np.log(disc_method_subset['pl_eqt']), label=f"{i}")

plt.legend()

plt.show()
```

```python
# Exoplanets found by imaging are more likely to be massive

ex_modified = ex_final[ex_final['pl_bmassj'] >5]
ex_modified

ex_modified['pl_eqt'].min()

final_answer = ex_modified[ex_modified['pl_eqt'] == ex_modified['pl_eqt'].min()]

final_answer
```

```python
# The coldest planet with a mass larger than five Jupiter masses is eps Ind A b

# Index = 566

# Equilibrium = 275 K

# Mass (Jupiter Masses) = 6.31

# Host Star Spectral Type = K5V: it is not a sun-like star
```

```python
plt.figure(figsize=(8,6))

plt.title("2019 Exoplanet Discoveries")

plt.xlabel("Mass Estimate (Jupiter Masses)")

plt.ylabel("Ideal Equilibrium Temperature")

disc_method_list = ex_final['discoverymethod'].unique()
# print(disc_method_list)

for i in disc_method_list:
    disc_method_subset = ex_final[ex_final['discoverymethod'] == i]
    plt.scatter(np.log(disc_method_subset['pl_bmassj']), np.log(disc_method_subset['pl_eqt']), label=f"{i}")

plt.scatter(np.log(final_answer['pl_bmassj']), np.log(final_answer['pl_eqt']), marker='*', s=100, label='Cold Giant Boi ❄️')

plt.legend()

plt.show()
```
